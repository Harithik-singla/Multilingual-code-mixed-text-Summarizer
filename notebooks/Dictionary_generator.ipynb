```python
import json
import pandas as pd
import re
from collections import Counter
from deep_translator import GoogleTranslator
from time import sleep

# Load existing dictionary
with open("hindi_english.json", "r", encoding="utf-8") as f:
    hindi_dict = json.load(f)

existing_hindi = set(hindi_dict.keys())

# Load dataset
df = pd.read_csv("train.csv", encoding='latin1')

TEXT_COLUMN = "text"  # change if needed

def extract_words(text):
    text = str(text)
    text = re.sub(r'[^ऀ-ॿ\s]', ' ', text)
    return text.split()

# 🔥 Count frequency
word_counter = Counter()

for sentence in df[TEXT_COLUMN]:
    words = extract_words(sentence)
    word_counter.update(words)

print("Total unique words:", len(word_counter))

# 🔥 Take top 2000 frequent words
most_common_words = [w for w, _ in word_counter.most_common(3000)]

# Remove existing + small garbage words
filtered_words = [
    w for w in most_common_words
    if w not in existing_hindi and len(w) > 2
]

# Limit to ~2000
filtered_words = filtered_words[:2000]

print("Words to translate:", len(filtered_words))

translator = GoogleTranslator(source='hi', target='en')

new_entries = {}

for word in filtered_words:
    try:
        english = translator.translate(word)

        if word not in hindi_dict and word not in new_entries:
            new_entries[word] = english.lower()
            print(f"Added: {word} -> {english}")

        sleep(0.2)

    except Exception as e:
        print(f"Error: {word} -> {e}")

# Merge
hindi_dict.update(new_entries)

# Save
with open("hindi_english_final.json", "w", encoding="utf-8") as f:
    json.dump(hindi_dict, f, ensure_ascii=False, indent=2)

print(f"Added {len(new_entries)} words")
print(f"Final size: {len(hindi_dict)}")
```
